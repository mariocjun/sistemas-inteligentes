# Laboratório de Busca e Inteligência Artificial (AIMA)

**Objetivo:** Explorar algoritmos de busca (A*) usando a implementação oficial de Russell & Norvig (`aima-python`), garantindo reprodutibilidade e ambiente determinístico.

## 1. Configuração do Ambiente e Reprodutibilidade
Vamos adicionar o repositório local ao nosso `PATH` do Python.

```python
import sys
import os
import random

# Adiciona o diretório lib/aima-python ao path para podermos importar seus módulos
aima_path = os.path.abspath(os.path.join(os.getcwd(), '..', 'lib', 'aima-python'))
if aima_path not in sys.path:
    sys.path.insert(0, aima_path)

# Fixando a semente para garantir resultados reproduzíveis (conforme regra GEMINI.md)
random.seed(42)

print(f"Diretório AIMA incluído no path: {aima_path}")
```

## 2. Formulando o Problema de Busca (PEAS / Grafo)
Vamos explorar o problema clássico de navegar pelo mapa da Romênia, de Arad a Bucareste, utilizando o módulo `search.py` do repositório.

```python
from search import GraphProblem, romania_map

# Instanciação do problema (Estado Inicial, Objetivo, Grafo de Transição)
romania_problem = GraphProblem('Arad', 'Bucharest', romania_map)

print("Cidades vizinhas a Arad e seus custos:")
print(romania_map.get('Arad'))
```

## 3. Busca Heurística com A*
O algoritmo A* (`astar_search`) utiliza a distância em linha reta até Bucareste como heurística $h(n)$ e a soma dos percursos como $g(n)$.

```python
from search import astar_search

# Executando a busca A*
node = astar_search(romania_problem)

print("Busca Concluída!")
print(f"-> Caminho ótimo encontrado: {node.solution()}")
print(f"-> Custo total do caminho g(n): {node.path_cost}")
```

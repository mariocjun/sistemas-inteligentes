## 1. Problema das N-Rainhas com GA
Vamos tentar posicionar 8 rainhas sem que elas se ataquem. A função de *fitness* (aptidão) conta o número de pares que **não** se atacam (máximo = 28 para 8 rainhas).

```python
def fitness_fn_nqueens(q):
    """Retorna o número de pares não-atacantes. O máximo para 8 rainhas é 28."""
    non_attacking = 0
    for col1 in range(len(q)):
        for col2 in range(col1 + 1, len(q)):
            if q[col1] != q[col2] and abs(col1 - col2) != abs(q[col1] - q[col2]):
                non_attacking += 1
    return non_attacking

# População inicial de 100 tabuleiros aleatórios. Gene pool é de 0 a 7 (as linhas)
gene_pool = list(range(8))
population = init_population(100, gene_pool, 8)

# Rodamos o GA. f_thres=28 significa parar se encontrarmos o ótimo.
solution = genetic_algorithm(population, fitness_fn_nqueens, gene_pool=gene_pool, f_thres=28, ngen=1000)

print("Melhor solução encontrada:", solution)
print("Aptidão da solução (Max=28):", fitness_fn_nqueens(solution))
```

**Objetivo:** Desconstruir a "caixa-preta" do algoritmo ID3 (Árvores de Decisão). Vamos calcular a Entropia e o Ganho de Informação passo a passo, matematicamente, antes de gerar qualquer árvore.

```python
import math

def calcular_entropia(probabilidades):
    """Calcula a Entropia H(S) = - sum(p_i * log2(p_i))"""
    entropia = 0.0
    for p in probabilidades:
        if p > 0:
            entropia -= p * math.log2(p)
    return entropia

# Exemplo: Uma moeda perfeitamente justa (50% Cara, 50% Coroa)
print("Entropia Moeda Justa:", calcular_entropia([0.5, 0.5]))

# Exemplo: Uma moeda viciada (99% Cara, 1% Coroa)
print("Entropia Moeda Viciada:", calcular_entropia([0.99, 0.01]))
```

## 1. Ganho de Informação
O **Ganho de Informação** nos diz qual atributo (ex: Chuva, Fome) reduz mais a incerteza (Entropia) sobre o nosso alvo (ex: Esperar mesa).

$$ IG(S, A) = H(S) - \sum_{v \in Valores(A)} \frac{|S_v|}{|S|} H(S_v) $$

```python
def ganho_de_informacao(entropia_sistema, subconjuntos):
    """
    entropia_sistema: Entropia antes de dividir (H(S)).
    subconjuntos: Lista de tuplas (peso_do_subconjunto, entropia_do_subconjunto)
    """
    entropia_apos_divisao = 0.0
    for peso, entropia in subconjuntos:
        entropia_apos_divisao += peso * entropia
        
    return entropia_sistema - entropia_apos_divisao

# Cenário hipotético: Dataset com 14 instâncias (9 Sim, 5 Não) -> Entropia = 0.940
H_S = 0.940

# Se dividirmos pelo atributo 'Chuva' (Forte/Fraca) geramos dois nós folhas hipotéticos:
# Nó Forte: 8 instâncias, Entropia = 0.811
# Nó Fraca: 6 instâncias, Entropia = 0.918
ganho = ganho_de_informacao(H_S, [(8/14, 0.811), (6/14, 0.918)])

print(f"Ganho de Informação ao usar o atributo 'Chuva': {ganho:.4f} bits")
```

## 2. Indução (A Magia do ID3)
O algoritmo ID3 não é uma caixa mágica. Ele simplesmente repete o cálculo acima para **todos os atributos disponíveis**, escolhe o atributo com maior Ganho de Informação, transforma ele num nó da árvore, e aplica o processo recursivamente para os galhos.

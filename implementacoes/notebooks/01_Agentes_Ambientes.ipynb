## 1. O Ambiente (Trivial Vacuum Environment)

## 2. Agente Reflexivo Simples
O Agente Reflexivo reage *apenas* à percepção atual (se o quadrado atual está sujo, ele aspira; senão, muda de sala). Ele não tem memória.

```python
# Criando o agente reflexivo
agente_reflexivo = ReflexVacuumAgent()

# Injetando o agente no ambiente
env.add_thing(agente_reflexivo)

# Habilitar log das ações do agente (Trace)
trace_agent(agente_reflexivo)

# Rodar a simulação por 10 passos
print("--- INICIANDO SIMULAÇÃO (AGENTE REFLEXIVO) ---")
env.run(10)
print("--- FIM DA SIMULAÇÃO ---")
```

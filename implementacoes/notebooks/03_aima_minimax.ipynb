```python
import sys
import os

aima_path = os.path.abspath(os.path.join(os.getcwd(), '..', 'lib', 'aima-python'))
if aima_path not in sys.path:
    sys.path.insert(0, aima_path)

from games import TicTacToe, minimax_decision, alphabeta_search, play_game, random_player
```

## 1. Definindo o Ambiente Adversarial
Inicializamos o estado do jogo. O `TicTacToe` implementa a estrutura formal do AIMA (ações possíveis, estado atual, função de utilidade e teste terminal).

```python
jogo = TicTacToe()
estado_inicial = jogo.initial

print("Ações possíveis no início do jogo:", jogo.actions(estado_inicial))
```

## 2. A Árvore de Decisão: Minimax vs Alfa-Beta
O `minimax_decision` navega toda a árvore, o que para jogos complexos é intratável. O `alphabeta_search` poda galhos matematicamente inúteis sem perder a otimização.

Vamos criar um jogador que usa Alfa-Beta e colocá-lo para jogar contra um jogador que faz movimentos aleatórios.

```python
def jogador_inteligente(jogo, estado):
    """Retorna o melhor movimento calculado pela poda alfa-beta."""
    return alphabeta_search(estado, jogo)

# O jogador 1 ('X') será nosso agente Alfa-Beta perfeito.
# O jogador 2 ('O') jogará aleatoriamente.
print("--- Partida: Inteligência Artificial (X) vs Aleatório (O) ---")
vencedor = play_game(jogo, dict(X=jogador_inteligente, O=random_player), verbose=True)

if vencedor == 1:
    print("\nResultado: A IA ('X') Venceu!")
elif vencedor == -1:
    print("\nResultado: O Aleatório ('O') Venceu! (Isto é matematicamente impossível para o Tic-Tac-Toe)")
else:
    print("\nResultado: Empate!")
```
